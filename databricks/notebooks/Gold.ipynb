### Create Gold table

```python
%sql
DROP TABLE IF EXISTS dev_hcahps2026.gold.complete_hcahps_data;
CREATE TABLE IF NOT EXISTS dev_hcahps2026.gold.complete_hcahps_data AS
---------------- Join hcahps_data and hospital_beds------------------
SELECT hcahps.*,
       beds.number_beds,
       beds.fiscal_year_begin_date as beds_start_report_date,
       beds.fiscal_year_end_date as beds_end_report_date    
FROM dev_hcahps2026.silver.hcahps_data as hcahps
LEFT JOIN dev_hcahps2026.silver.hospital_beds as beds
ON hcahps.provider_ccn = beds.provider_ccn
WHERE beds.number_beds IS NOT NULL
```
