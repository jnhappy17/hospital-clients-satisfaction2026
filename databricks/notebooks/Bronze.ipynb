### Create table hcahps_data

```python
%sql
DROP TABLE IF EXISTS dev_hcahps2026.bronze.hcahps_data;
CREATE TABLE IF NOT EXISTS dev_hcahps2026.bronze.hcahps_data
(
    facility_id STRING,
    facility_name STRING,
    address STRING,
    city STRING,
    state VARCHAR(2),
    zipcode VARCHAR(10),
    county_parish STRING,
    phone_number VARCHAR(20),
    hcahps_measure_id STRING,
    hcahps_question STRING,
    hcahps_answer_description STRING,
    hcahps_answer_percent STRING,
    num_completed_surveys STRING,
    survey_response_rate_percent STRING,
    start_date VARCHAR(10),
    end_date VARCHAR(10)
)
```

### Create table hospital_beds

### Load table hospital_beds
